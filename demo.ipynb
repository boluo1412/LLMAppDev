```python
import os
from dotenv import load_dotenv,find_dotenv
_=load_dotenv(find_dotenv())
```

```python
from openai import OpenAI
client=OpenAI(
    api_key=os.environ.get("OPEN_AI_KEY"),
    base_url="https://api.deepseek.com"
)
```

```python
def send_messages(prompt):
    message=[{"role":"user","content":prompt}]
    return message
```

```python
def get_completion(prompt,model="deepseek-v4-flash",temperature=0):
    response=client.chat.completions.create(model=model,messages=send_messages(prompt),temperature=temperature)
    if len(response.choices) > 0:
        return response.choices[0].message.content
    return "生成答案时出错"
```

```python
get_completion("你好，介绍一下你自己吧")
```

```python
prompt = f"""
请生成包括书名、作者和类别的三本虚构的、非真实存在的中文书籍清单，\
并以 JSON 格式提供，其中包含以下键:book_id、title、author、genre。
"""
response = get_completion(prompt)
print(response)

```
